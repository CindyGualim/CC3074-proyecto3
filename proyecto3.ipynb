***Implementación Base***

```python
def mtf_access(lst: list, element) -> tuple[int, list]:

    lst = lst.copy()
    pos = lst.index(element)          
    cost = pos + 1                   
   
    while pos > 0:
        lst[pos], lst[pos - 1] = lst[pos - 1], lst[pos]
        pos -= 1
    return cost, lst


def run_mtf(initial_list: list, requests: list, verbose: bool = True) -> int:

    lst = initial_list.copy()
    total_cost = 0

    if verbose:
        col = 22
        print(f"{'Config. actual':<{col}} {'Solicitud':>10} {'Costo':>7}  {'Config. resultante'}")
        print("-" * 70)

    for req in requests:
        config_before = lst.copy()
        cost, lst = mtf_access(lst, req)
        total_cost += cost
        if verbose:
            print(f"{str(config_before):<{col}} {str(req):>10} {cost:>7}  {lst}")

    if verbose:
        print("-" * 70)
        print(f"{'Costo total de accesos':>40}: {total_cost}")

    return total_cost



def imtf_access(lst: list, requests: list, idx: int) -> tuple[int, list]:

    lst = lst.copy()
    element = requests[idx]
    pos = lst.index(element)        # 0-indexed
    cost = pos + 1                  # 1-indexed

    # look-ahead: los próximos (pos) elementos en la secuencia
    look_ahead = requests[idx + 1 : idx + 1 + pos]  # next (pos-1) => pos elems after idx?
    # Según el enunciado: "próximos i-1 elementos" donde i = posición 1-indexed
    look_ahead = requests[idx + 1 : idx + pos]       # i-1 elementos hacia adelante

    if element in look_ahead:
        # mover al frente
        while pos > 0:
            lst[pos], lst[pos - 1] = lst[pos - 1], lst[pos]
            pos -= 1

    return cost, lst


def run_imtf(initial_list: list, requests: list, verbose: bool = True) -> int:

    lst = initial_list.copy()
    total_cost = 0

    if verbose:
        col = 22
        print(f"{'Config. actual':<{col}} {'Solicitud':>10} {'Costo':>7}  {'Config. resultante'}")
        print("-" * 70)

    for i, req in enumerate(requests):
        config_before = lst.copy()
        cost, lst = imtf_access(lst, requests, i)
        total_cost += cost
        if verbose:
            print(f"{str(config_before):<{col}} {str(req):>10} {cost:>7}  {lst}")

    if verbose:
        print("-" * 70)
        print(f"{'Costo total de accesos (IMTF)':>40}: {total_cost}")

    return total_cost
```

***Ejercicio 3 – Secuencia de mínimo costo (20 solicitudes)***

Se busca la secuencia de 20 solicitudes sobre `[0,1,2,3,4]` con el menor
costo total posible en MTF.

**Razonamiento:** El costo mínimo se obtiene cuando el elemento solicitado
siempre está al frente de la lista (costo = 1 cada vez).
Esto ocurre si repetimos el mismo elemento 20 veces, o si repetimos la
misma secuencia en el mismo orden en que MTF deja la lista.
La secuencia óptima trivial es repetir el elemento 0 (que ya está al frente).


```python
from itertools import product as iproduct

config_3 = [0, 1, 2, 3, 4]

best_seq   = [0] * 20
best_cost  = run_mtf(config_3, best_seq, verbose=False)

print(f"Secuencia de mínimo costo: {best_seq}")
print(f"Costo total mínimo       : {best_cost}")
print()
print("Verificación detallada:")
run_mtf(config_3, best_seq)
```

***Ejercicio 4 – Peor caso (20 solicitudes)***

**Razonamiento:** El peor caso para MTF es cuando cada solicitud apunta al
elemento que se encuentra al **final** de la lista actual, forzando el
mayor costo posible en cada acceso.

Para una lista de 5 elementos, el adversario siempre pide el elemento en la
última posición (costo = 5 la primera vez, luego varía según cómo MTF
reordena). Se puede explorar sistemáticamente.


```python
import random

config_4 = [0, 1, 2, 3, 4]


def worst_case_greedy(initial_list: list, n: int) -> list:

    lst   = initial_list.copy()
    seq   = []
    for _ in range(n):
        worst_elem = lst[-1]
        seq.append(worst_elem)
        _, lst = mtf_access(lst, worst_elem)
    return seq

worst_seq  = worst_case_greedy(config_4, 20)
worst_cost = run_mtf(config_4, worst_seq, verbose=False)

print(f"Secuencia del peor caso: {worst_seq}")
print(f"Costo total peor caso  : {worst_cost}")
print()
print("Verificación detallada:")
run_mtf(config_4, worst_seq)
```

***Ejercicio 6 – Algoritmo IMTF (Improved Move-to-Front)***

Se aplica IMTF sobre:
- **Mejor caso de MTF** (secuencia del Ejercicio 3)
- **Peor caso de MTF** (secuencia del Ejercicio 4)

**Regla IMTF:** después de acceder al elemento en posición *i* (1-indexed),
se mueve al frente **solo si** aparece en los próximos *i−1* elementos
de la secuencia de solicitudes restante.


```python
config_6 = [0, 1, 2, 3, 4]

#  IMTF sobre el MEJOR CASO de MTF 
print("=" * 70)
print("EJERCICIO 6a – IMTF sobre el MEJOR CASO de MTF")
print(f"Secuencia: {best_seq}")
print("=" * 70)
total_imtf_best = run_imtf(config_6, best_seq)
print()
print(f"MTF  costo total (mejor caso): {best_cost}")
print(f"IMTF costo total (mejor caso): {total_imtf_best}")
```

```python
#  IMTF sobre el PEOR CASO de MTF 
print("=" * 70)
print("EJERCICIO 6b – IMTF sobre el PEOR CASO de MTF")
print(f"Secuencia: {worst_seq}")
print("=" * 70)
total_imtf_worst = run_imtf(config_6, worst_seq)
print()
print(f"MTF  costo total (peor caso) : {worst_cost}")
print(f"IMTF costo total (peor caso) : {total_imtf_worst}")
```
